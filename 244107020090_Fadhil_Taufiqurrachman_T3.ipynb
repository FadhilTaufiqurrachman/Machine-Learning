```python
# Import Library
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

# Load Data
df = pd.read_csv(BASE_PATH + 'wbc.csv')
print("Data Mentah Awal :")
display(df.head())
```

**1. Memisahkan Variable**

```python
df_clean = df.drop(columns=['id', 'Unnamed: 32'])

print("Struktur Data etelah menghapus kolom yang tidak berguna :")
df_clean.info()
```

**2. Encoding Kolom Diagnosis**

```python
le = LabelEncoder()
df_clean['diagnosis'] = le.fit_transform(df_clean['diagnosis'])

print("Kolom 'diagnosis' setelah di-encode (0 = Benign, 1 = Malignant):")
print(df_clean['diagnosis'].value_counts())
display(df_clean[['diagnosis']].head())
```

**3. Standarisasi**

```python
X = df_clean.drop(columns=['diagnosis'])
y = df_clean['diagnosis']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

print("Statistik deskriptif data setelah standarisasi:")
display(X_scaled_df.describe().round(3).loc[['mean', 'std']])
```

**4. Stratified Split**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled_df, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print(f"Bentuk (Shape) Data Latih (X_train) : {X_train.shape}")
print(f"Bentuk (Shape) Data Uji (X_test)    : {X_test.shape}")

print(f"\nJumlah Data Latih: {X_train.shape[0]}")
print(f"Jumlah Data Uji: {X_test.shape[0]}")

print("\nJumlah kelas pada Data Latih (y_train):")
print(y_train.value_counts())

print("\nJumlah kelas pada Data Uji (y_test):")
print(y_test.value_counts())
```
